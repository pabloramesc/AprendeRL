# Study n-step SARSA

n-step SARSA trades one-step bias for a longer sampled return. The experiment uses deterministic `FrozenLake-v1` so the learned table can be inspected directly.

## Defining update

$$G_{t:t+n}=\sum_{k=0}^{n-1}\gamma^kR_{t+k+1}+\gamma^nQ(S_{t+n},A_{t+n})$$

```python
import gymnasium as gym
import matplotlib.pyplot as plt
import numpy as np

from aprenderl import NStepSARSA, NStepSARSAConfig
from aprenderl.utils import evaluate_policy

ENV_ID = "FrozenLake-v1"
```

```python
env = gym.make(ENV_ID, is_slippery=False)
config = NStepSARSAConfig(
    n_steps=4,
    exploration_steps=3_000,
    seed=7,
)
agent = NStepSARSA(env, config=config)
agent.learn(5_000, progress_bar=False)
```

```python
print("Q-table shape:", agent.q_table.shape)
print("Greedy policy:")
print(agent.q_table.argmax(axis=1).reshape(4, 4))
print("Updates:", agent.num_updates)
```

```python
returns = np.asarray(agent.episode_returns)
plt.figure(figsize=(8, 3))
plt.plot(returns, alpha=0.5)
plt.xlabel("Episode")
plt.ylabel("Return")
plt.title("n-step SARSA training")
plt.show()
```

```python
evaluation_env = gym.make(ENV_ID, is_slippery=False)
result = evaluate_policy(agent, evaluation_env, episodes=20)
print("Mean return:", result.mean_return)
evaluation_env.close()
env.close()
```
