# Train Value Iteration

Value Iteration plans directly from the finite transition model exposed by Gymnasium's FrozenLake environment.

## Bellman operation

$$V_{k+1}(s)=\max_a\sum_{s',r}p(s',r\mid s,a)\left[r+\gamma V_k(s')\right].$$

Here $V_k$ is the value estimate at sweep $k$, $p$ the environment model, and $\gamma$ the discount factor.

```python
import gymnasium as gym
import matplotlib.pyplot as plt

from aprenderl import DynamicProgrammingConfig, ValueIteration
from aprenderl.utils import evaluate_policy

ENV_ID = "FrozenLake-v1"
```

```python
env = gym.make(ENV_ID, is_slippery=True)
config = DynamicProgrammingConfig(seed=7)

agent = ValueIteration(env, config=config)
agent.learn(total_timesteps=1_000, progress_bar=False)
env.close()

print("Converged:", agent.converged)
print("Planning sweeps:", agent.num_timesteps)
print(agent.policy.reshape(4, 4))
```

```python
plt.imshow(agent.values.reshape(4, 4), cmap="viridis")
plt.colorbar(label="State value")
plt.title("Value Iteration values")
plt.show()
```

## Watch the planned policy

This opens a window and runs 5 episodes using deterministic actions.

```python

evaluation_env = gym.make(
    ENV_ID, render_mode="human", is_slippery=True
)
try:
    result = evaluate_policy(
        agent, evaluation_env, episodes=5, deterministic=True, seed=10
    )
finally:
    evaluation_env.close()

print("Episode returns:", result.returns)
print(f"Mean return: {result.mean_return:.1f} +/- {result.return_std:.1f}")
```
