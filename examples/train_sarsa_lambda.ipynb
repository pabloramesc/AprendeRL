# Train SARSA(lambda)

SARSA(lambda) assigns each TD error backward through eligibility traces. The experiment uses deterministic `FrozenLake-v1` so the learned table can be inspected directly.

## Defining update

$$E_t=\gamma\lambda E_{t-1}+\mathbf{1}\{S_t,A_t\}$$

```python
import gymnasium as gym
import matplotlib.pyplot as plt
import numpy as np

from aprenderl import SARSALambda, SARSALambdaConfig
from aprenderl.utils import evaluate_policy

ENV_ID = "FrozenLake-v1"
```

```python
env = gym.make(ENV_ID, is_slippery=False)
config = SARSALambdaConfig(
    trace_decay=0.9,
    exploration_steps=3_000,
    seed=7,
)
agent = SARSALambda(env, config=config)
agent.learn(5_000, progress_bar=False)
```

```python
print("Q-table shape:", agent.q_table.shape)
print("Greedy policy:")
print(agent.q_table.argmax(axis=1).reshape(4, 4))
print("Updates:", agent.num_updates)
```

```python
returns = np.asarray(agent.episode_returns)
plt.figure(figsize=(8, 3))
plt.plot(returns, alpha=0.5)
plt.xlabel("Episode")
plt.ylabel("Return")
plt.title("SARSA(lambda) training")
plt.show()
```

```python
evaluation_env = gym.make(ENV_ID, is_slippery=False)
result = evaluate_policy(agent, evaluation_env, episodes=20)
print("Mean return:", result.mean_return)
evaluation_env.close()
env.close()
```
