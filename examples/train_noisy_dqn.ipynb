# Train NoisyNet DQN

NoisyNet DQN learns parameter-space noise instead of relying on epsilon-greedy exploration. This notebook trains it on `CartPole-v1` and plots its learning curve.

## Defining idea

$$W=\mu_W+\sigma_W\odot\varepsilon_W.$$

Here $\mu_W$ and $\sigma_W$ are learned parameter means and noise scales, $\varepsilon_W$ is sampled noise, and $\odot$ denotes element-wise multiplication.

```python
import gymnasium as gym
import matplotlib.pyplot as plt
import numpy as np

from aprenderl import NoisyDQN, NoisyDQNConfig
from aprenderl.utils import evaluate_policy

ENV_ID = "CartPole-v1"
```

```python
env = gym.make(ENV_ID)
config = NoisyDQNConfig(
    noise_sigma=0.5,
    buffer_size=10_000,
    batch_size=64,
    learning_starts=500,
    target_update_interval=250,
    exploration_steps=2_000,
    seed=7,
)

agent = NoisyDQN(env, config=config, device="cpu")
agent.learn(total_timesteps=5_000, progress_bar=False)
env.close()
```

```python
returns = np.asarray(agent.episode_returns)
window = min(10, len(returns))
moving_average = np.convolve(
    returns, np.ones(window) / window, mode="valid"
)

plt.figure(figsize=(8, 4))
plt.plot(returns, alpha=0.35, label="Episode return")
plt.plot(
    np.arange(window - 1, len(returns)),
    moving_average,
    label=f"{window}-episode average",
)
plt.xlabel("Episode")
plt.ylabel("Return")
plt.title(f"NoisyNet DQN training on {ENV_ID}")
plt.legend()
plt.grid(alpha=0.2)
plt.show()
```

## Watch the trained policy

This opens a window and runs 5 episodes using deterministic actions.

```python

evaluation_env = gym.make(ENV_ID, render_mode="human")
try:
    result = evaluate_policy(
        agent, evaluation_env, episodes=5, deterministic=True
    )
finally:
    evaluation_env.close()

print("Episode returns:", result.returns)
print(f"Mean return: {result.mean_return:.1f} +/- {result.return_std:.1f}")
```
